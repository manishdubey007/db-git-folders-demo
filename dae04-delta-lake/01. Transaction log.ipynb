#### Understanding Delta Lake Transaction Log
Understand the cloud storage directory structure of delta lake tables

####0. Create a new Schema under the demo catalog for this section of course(delta_lake)

```python
%sql
CREATE CATALOG IF NOT EXISTS demo
MANAGED LOCATION 'abfss://demo@databrickscourseextmandl.dfs.core.windows.net/'
```

```python
%sql
DROP SCHEMA IF EXISTS demo.delta_lake
```

```python
%sql
CREATE SCHEMA IF NOT EXISTS demo.delta_lake
    MANAGED LOCATION 'abfss://demo@databrickscourseextmandl.dfs.core.windows.net/delta_lake'

```

```python
%sql
CREATE TABLE IF NOT EXISTS demo.delta_lake.companies
(
    company_name STRING,
    founded_date DATE,
    country STRING
)
```

```python
%sql
DESCRIBE EXTENDED demo.delta_lake.companies
```

####2. Insert Some Data
