####History and time travel
- Query Delta Lake Table History
- Query previous version of the data
- query data from specific time 
- Restore data to a specific version

##### 1. Query Delta Lake Table History

```python
%sql
DESCRIBE HISTORY demo.delta_lake.companies;
```

#####2. Query previous version of the data


```python
%sql
SELECT * FROM demo.delta_lake.companies
VERSION AS OF 2;
```

#####3. Query data from specific time

```python
%sql
SELECT * FROM demo.delta_lake.companies
TIMESTAMP AS OF ('2026-08-23T12:01:17.000+00:00');
```

#####4. Restore data to a specific version

```python
%sql
RESTORE TABLE demo.delta_lake.companies VERSION AS OF 1;
    
```

```python
%sql
SELECT * FROM demo.delta_lake.companies
```
