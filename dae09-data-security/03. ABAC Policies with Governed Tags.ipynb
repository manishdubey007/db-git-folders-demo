###Data-Level Security using ABAC(Attribute Based Access Control) Policies (Governed Tags) in Unity Catalog
ABAC addresses this by introducing a centralized model where:
- Data is tagged using governed attributes
- Policies are defined based on these attributes
- Policies are applied automatically across multiple objects

The objective is to:
    - Classify sensitive dtat using governed tags
    - Define reusable policies based on those tags


####1. Set up the schema for the table

```sql
USE CATALOG demo;
USE SCHEMA data_security;
```

```sql
CREATE OR REPLACE TABLE sales_abac(
    id INT,
    region STRING,
    email STRING,
    revenue INT
)
```

```sql
INSERT INTO sales_abac VALUES
(1, 'UK', 'john.doe@gmail.com', 100),
(2, 'US', 'jane.smith@gmail.com', 200),
(3, 'UK', 'bob.jones@gmail.com', 300),
(4, 'US', 'sara.williams@gmail.com', 400);
```

```sql
SELECT * FROM sales_abac
```

#### Crate UDFs for row filter and column mask
- fn_filter_region
- fn_mask_email

#### Create Governed Tags in catalog
- pii = email
- access_type = region

#### Create policy(could be at catalog/table) level
1. policy_filter_region
2. policy_mask_email

#### Attach Tags To the Table (in Catalog)
 - sales_abac_region
 - sales_abac_email

#### Create another table to demonstrate scalability of ABAC implementation

```sql
CREATE OR REPLACE TABLE customers_abac (
    customer_id INT,
    customer_name STRING,
    email STRING,
    region STRING
)
```

```sql
ALTER TABLE customers_abac
ALTER COLUMN email
SET TAGS ('pii' = 'email');
```

```sql
ALTER TABLE customers_abac
ALTER COLUMN region
SET TAGS ('access_type' = 'region');
```

```sql
INSERT INTO customers_abac VALUES
(1, 'John Doe', 'john.doe@gmail.com', 'UK'),
(2, 'Jane Smith', 'jane.smith@gmail.com', 'US'),
(3, 'Bob Jones', 'bob.jones@gmail.com', 'UK'),
(4, 'Sara Williams', 'sara.williams@gmail.com', 'US')
```
