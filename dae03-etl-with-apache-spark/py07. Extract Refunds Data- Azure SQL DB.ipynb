#### Loading refund data
1. Read returns data from JDBC
2. Write data in bronze schema

#### 1. create connection - asql_gimobox_db_conn

```python
df = spark.read.format("jdbc").option("url", "jdbc:sqlserver://gizmobox-serv.database.windows.net:1433;database=gizmobox-db").option("dbtable", "refunds").option("user", "gizmoboxadm").option("password", "Infy@123").load()

display(df)

## https://spark.apache.org/docs/latest/api/python/reference/pyspark.sql/api/pyspark.sql.DataFrameReader.jdbc.html
```

####2. Write data in bronze layer

```python
df.write.mode("overwrite").saveAsTable("gizmobox.bronze.py_refunds")
```
