####Extract data from the orders JSON File
1. Query order files using JSON
2. Query order files using TEXT
3. Create order view in bronze schema

#### 1. Query order files using JSON

```python
%sql
SELECT * FROM json.`/Volumes/gizmobox/landing/operational_data/orders/` LIMIT 10
```

#### 2. Query order files using TEXT

```python
%sql
SELECT * FROM text.`/Volumes/gizmobox/landing/operational_data/orders/` LIMIT 10
```

#### 3. Create order view in bronze schema

```python
%sql
CREATE OR REPLACE VIEW gizmobox.bronze.v_orders 
AS 
SELECT
    * ,
    _metadata.file_path AS file_path
FROM text.`/Volumes/gizmobox/landing/operational_data/orders/`;
    
```

```python
%sql
SELECT * FROM gizmobox.bronze.v_orders LIMIT 10;
```
