###Extract data from the Address Files
1. Demo of limitation of CSV file format in SELECT statement
2. Use read-files function to overcome the limitations
3. Create address view in the bronze layer

#### 1. Demo of limitation of CSV file format in SELECT statement

```python
%sql
SELECT * FROM csv.`/Volumes/gizmobox/landing/operational_data/addresses`
```

#### 2. Use read-files function to overcome the limitations
https://docs.databricks.com/aws/en/sql/language-manual/functions/read_files

```python
%sql
SELECT * FROM read_files('/Volumes/gizmobox/landing/operational_data/addresses',
                    format => 'csv',
                    delimiter => '\t',
                    header => 'true')
```

```python
%sql
CREATE OR REPLACE VIEW gizmobox.bronze.v_addresses AS
SELECT * FROM read_files('/Volumes/gizmobox/landing/operational_data/addresses',
                    format => 'csv',
                    delimiter => '\t',
                    header => 'true')
```

```python
%sql
SELECT * FROM gizmobox.bronze.v_addresses
```
