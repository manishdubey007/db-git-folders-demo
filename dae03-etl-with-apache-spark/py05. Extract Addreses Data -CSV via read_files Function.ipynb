###Extract data from the Address Files
1. Read addresses data
3. Create address table in the bronze layer

#### 1. Demo of limitation of CSV file format in SELECT statement

```python
df = (
    spark.read.format("csv")
    .option("delimiter", "\t")
    .option("header", "true")
    .load("/Volumes/gizmobox/landing/operational_data/addresses")
    )

display(df)
```

####2. Create table in bronze layer

```python
df.writeTo("gizmobox.bronze.py_addresses").createOrReplace()
```

```python
%sql
SELECT * FROM gizmobox.bronze.py_addresses
```
