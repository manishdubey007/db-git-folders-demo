#### Transform Addresses Data
- Create 1 record for each customer with 2 sets of address columns, 1 for shipping address and 1 for billing address
- Write transformed data in to the silvaer layer

#### 1. Create 1 record for each customer with 2 sets of address columns, 1 for shipping address and 1 for billing address
- https://docs.databricks.com/aws/en/sql/language-manual/sql-ref-syntax-qry-select-pivot


```python
%sql
SELECT
    *
FROM gizmobox.bronze.v_addresses
```

```python
%sql
SELECT
    *
FROM (
    SELECT
        customer_id,
        address_type,
        address_line_1,
        city,
        state,
        postcode
    FROM gizmobox.bronze.v_addresses
)
PIVOT(
    MAX(address_line_1) AS address_line_1,
    MAX(city) AS address_city,
    MAX(state) AS address_state,
    MAX(postcode) AS address_postcode
    FOR address_type IN ('billing', 'shipping')
);
```

#### 2. Write transformed data in to the silvaer layer

```python
%sql
CREATE TABLE gizmobox.silver.addresses
AS
SELECT
    *
FROM (
    SELECT
        customer_id,
        address_type,
        address_line_1,
        city,
        state,
        postcode
    FROM gizmobox.bronze.v_addresses
)
PIVOT(
    MAX(address_line_1) AS address_line_1,
    MAX(city) AS address_city,
    MAX(state) AS address_state,
    MAX(postcode) AS address_postcode
    FOR address_type IN ( 'shipping', 'billing')
);
```

```python
%sql
SELECT * FROM gizmobox.silver.addresses
```
