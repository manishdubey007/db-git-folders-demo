####Query Orders Data - JSON strings
- Extract Top Level Column Values
- Extract Array Elements
- Extract Nested Column Values
- Cast columns to a specific datatype

#### 1.Extract Top Level Column Values


```python
%sql
SELECT
    value:order_id,
    value
FROM gizmobox.bronze.v_orders
```

```python
%sql
SELECT
    get_json_object(value, '$.order_id') AS order_id,
    get_json_object(value, '$.customer_id') AS customer_id,
    get_json_object(value, '$.order_date') AS order_date,
    get_json_object(value, '$.transaction_timestamp') AS transaction_timestamp,
    get_json_object(value, '$.total_amount') AS total_amount,
    get_json_object(value, '$.payment_method') AS payment_method,
    get_json_object(value, '$.order_status') AS order_status,
    get_json_object(value, '$.items') AS items,
    get_json_object(value, '$.file_path') AS file_path,
    *
FROM gizmobox.bronze.v_orders
```

#### 2. Extract Array Elements

```python
%sql
SELECT
    value:items[0] AS items_value_0,
    value:items[1] AS items_value_1,
    value
FROM gizmobox.bronze.v_orders
```

####3. Extract Nested Column Values
<column_name>:<extraction_path>[Index].<leaf_node>

```python
%sql
SELECT
    value:items[0].item_id AS items_id_0,
    value:items[1].item_id AS items_id_1,
    value
FROM gizmobox.bronze.v_orders
```

####4. Cast columns to a specific datatype

```python
%sql
SELECT
    value:items[0].item_id::INTEGER AS items_id_0,
    value:items[1].item_id::INTEGER AS items_id_1,
    value
FROM gizmobox.bronze.v_orders
```
