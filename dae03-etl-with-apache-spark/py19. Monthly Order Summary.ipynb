#### Monthly Order Summary for each customer per month

- Total orders
- Total items bought
- Total amount spent

#### 1. Total orders

```python
from pyspark.sql import functions as F

orders_df = spark.read.table("gizmobox.silver.py_orders").withColumn("order_month", F.date_format("transaction_timestamp", "yyyy-MM"))

total_orders_df = (
    orders_df
    .groupBy("customer_id", "order_month")
    .agg(
        F.countDistinct("order_id").alias("total_orders"),
        F.sum("quantity").alias("total_items_bought"),
        F.sum(F.col("quantity") * F.col("price")).alias("total_amount_spent")
        )
)
display(total_orders_df)
```

####3. Total amount spent

```python
total_orders_df.writeTo("gizmobox.gold.py_order_summary_monthly").createOrReplace()
```
