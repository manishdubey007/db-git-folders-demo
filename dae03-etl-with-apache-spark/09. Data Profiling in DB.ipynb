### Data Profiling in DB
- Profile data using UI
- Profile data using DBUTILS Package (dbutils.data.summarize method)
- Profile data manually 
    - COUNT
    - COUNT_IF
    - MIN
    - MAX
    - WHERE Clause

#### Profile data using UI

```python
%sql
SELECT * FROM gizmobox.bronze.v_customers;
```

####Profile data using DBUTILS Package (dbutils.data.summarize method)

```python
df = spark.table('gizmobox.bronze.v_customers')
dbutils.data.summarize(df)
```

#### Profile data manually
- COUNT
- COUNT_IF
- MIN
- MAX
- WHERE Clause

```python
%sql
SELECT 
    COUNT(*), 
    COUNT(customer_id),
    COUNT(email),
    COUNT(telephone)
FROM gizmobox.bronze.v_customers
```

```python
%sql
SELECT 
    COUNT(*), 
    COUNT_IF(customer_id IS NULL),
    COUNT_IF(email IS NULL),
    COUNT_IF(telephone IS NULL)
FROM gizmobox.bronze.v_customers
```

```python
%sql
SELECT 
    COUNT(*)
FROM gizmobox.bronze.v_customers
WHERE customer_id IS NULL;
```

```python
%sql
SELECT 
    COUNT(*),
    COUNT(DISTINCT customer_id) as unique_customer_id
FROM gizmobox.bronze.v_customers
WHERE customer_id IS NOT NULL;
```
