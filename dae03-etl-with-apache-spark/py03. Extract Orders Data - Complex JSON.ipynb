####Extract data from the orders JSON File
1. Query order files using JSON
2. Query order files using TEXT
3. Create order table in bronze schema

#### 1. Query order files using JSON

```python
df = spark.read.json("/Volumes/gizmobox/landing/operational_data/orders/*")
display(df)
```

#### 2. Query order files using TEXT

```python
df = spark.read.format("text").load("/Volumes/gizmobox/landing/operational_data/orders/*")
display(df)
```

#### 3. Create order view in bronze schema

```python

df.writeTo("gizmobox.bronze.py_orders").createOrReplace()
```

```python
%sql
SELECT * FROM gizmobox.bronze.py_orders LIMIT 10;
```
