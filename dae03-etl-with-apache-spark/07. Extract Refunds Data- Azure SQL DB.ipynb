#### Configure Access to Azure SQL Databases via Lakehouse Federation
1. create connection  - asql_gimobox_db_conn
2. create foreign catalog - asql_gizmobox_db_catalog

#### 1. create connection - asql_gimobox_db_conn

```python
%sql
CREATE CONNECTION IF NOT EXISTS asql_gimobox_db_conn
  TYPE sqlserver
  OPTIONS ( 
    host 'gizmobox-serv.database.windows.net',
    port '1433',
    user 'gizmoboxadm',
    password 'Infy@123'
)
```

####2. create foreign catalog - asql_gizmobox_db_catalog

```python
%sql
CREATE FOREIGN CATALOG IF NOT EXISTS asql_gizmobox_db_catalog
    USING CONNECTION asql_gimobox_db_conn
    OPTIONS ( 
        database 'gizmobox-db'
    )
```

```python
%sql SHOW CATALOGS;
USE CATALOG asql_gizmobox_db_catalog;
SHOW SCHEMAS;
USE SCHEMA dbo;
SHOW TABLES;
SELECT * FROM refunds


```

```python
%sql
SELECT 
    * 
FROM refunds r 
INNER JOIN gizmobox.bronze.payments p
ON r.payment_id = p.payment_id
```

#### Creating table 

```python
%sql
CREATE SCHEMA IF NOT EXISTS gizmobox.bronze;

CREATE TABLE IF NOT EXISTS gizmobox.bronze.refunds AS
SELECT
    refund_id,
    payment_id,
    refund_timestamp,
    refund_amount,
    refund_reason
FROM asql_gizmobox_db_catalog.dbo.refunds;
```
