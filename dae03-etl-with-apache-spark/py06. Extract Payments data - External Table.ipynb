#### 2. Create external table 

```python
payements_schema = 'payment_id INTEGER, order_id INTEGER, payment_timstamp TIMESTAMP, payement_status INTEGER, payment_method STRING '
```

```python
df = (
    spark.read.format("csv")
    .option("delimiter", ",")
    .schema(payements_schema)
    .load("abfss://gizmobox@databrickscourseextmandl.dfs.core.windows.net/landing/external_data/payments/")
)

display(df)
## https://spark.apache.org/docs/latest/api/python/reference/pyspark.sql/api/pyspark.sql.DataFrameReader.schema.html
```

```python
from pyspark.sql.types import StructType, StructField, StringType, IntegerType, TimestampType

py_payments_schema = StructType([
    StructField('payment_id', IntegerType()),
    StructField('order_id', IntegerType()),
    StructField('payment_timstamp', TimestampType()),
    StructField('payement_status', IntegerType()),
    StructField('payment_method', StringType())
])
```

```python
py_df = (
    spark.read.format("csv")
    .option("delimiter", ",")
    .schema(py_payments_schema)
    .load("abfss://gizmobox@databrickscourseextmandl.dfs.core.windows.net/landing/external_data/payments/")
)

display(py_df )
```

```python
py_df.writeTo('gizmobox.bronze.py_payments').createOrReplace()
 
```
