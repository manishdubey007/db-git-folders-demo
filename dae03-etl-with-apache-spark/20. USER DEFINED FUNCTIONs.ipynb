#### USER DEFINED FUNCTIONS
- UDF in spark are custom functions created by users to extend the capbilities of Spark-SQL/Py-Spark
- UDFs allow us to perform calculations or transformations to apply business logic that are not possible with in-builts functions
- You define the function once and use it across multiple queries 
- SQL UDFs are recommended over python UDFs due to better optimization

#### Syntax
------------------------------------------------------------------------------------------
`CREATE FUNCTION catalog_name.schema_name.udf_name(param_name datatype)
RETURNS return_type
RETURN expression;`

```python
%sql
CREATE OR REPLACE FUNCTION gizmobox.default.get_fullname(first_name STRING, last_name STRING) 
RETURNS STRING 
RETURN CONCAT(initcap(first_name), ' ', initcap(last_name));
```

```python
%sql
SELECT gizmobox.default.get_fullname('Manish', 'Patel')
```

```python
%sql
DESC FUNCTION EXTENDED gizmobox.default.get_fullname;
```
