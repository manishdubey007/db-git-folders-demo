#### Transform Membership Data
- Extract customer_id from the file path
- Write transformed data in silver schema


####1. Extract customer_id from the file path
- https://docs.databricks.com/aws/en/sql/language-manual/functions/regexp_extract
- https://regexr.com/
- https://docs.oracle.com/javase/8/docs/api/java/util/regex/Pattern.html

```python
from pyspark.sql import functions as F

df = spark.read.table("gizmobox.bronze.py_memberships")
extracted_df = df.select(
    F.regexp_extract(df.path, ".*/([0-9]+)\\.png$", 1).cast("Integer").alias("customer_id"),
    df.content.alias("membership_card")
    )
display(extracted_df)
```

#### 2. Write transformed data in silver schema

```python
extracted_df.writeTo("gizmobox.silver.py_memberships").createOrReplace()
```
