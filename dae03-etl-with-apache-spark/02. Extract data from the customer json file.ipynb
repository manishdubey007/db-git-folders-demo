### Extract data from the customer json file
- Querying single file
- Querying files with wild character
- Querying directory
- Including input file path or metdata
- Registering files in unity catalog using views 

#### 1. Querying single file

```python
%fs ls /Volumes/gizmobox/landing/operational_data/customers/
```

```python
%sql
SELECT * FROM json.`dbfs:/Volumes/gizmobox/landing/operational_data/customers/customers_2024_10.json` LIMIT 10
```

#### 2. Reading multiple files

```python
%sql
SELECT * FROM json.`dbfs:/Volumes/gizmobox/landing/operational_data/customers/customers_2024_*.json`;
```

#### 3. Entire folder

```python
%sql
SELECT * FROM json.`dbfs:/Volumes/gizmobox/landing/operational_data/customers/`
```

#### 4. Select file metadat

```python
%sql
SELECT 
    input_file_name() AS file_path,  -- will ne depricated
    _metadata.file_path,
    *
FROM json.`dbfs:/Volumes/gizmobox/landing/operational_data/customers/`
```

#### 5. Registering files in view

```python
%sql
CREATE OR REPLACE VIEW gizmobox.bronze.v_customers AS
SELECT 
    *,
    _metadata.file_path
FROM json.`dbfs:/Volumes/gizmobox/landing/operational_data/customers/`;
```

```python
%sql
SELECT * FROM gizmobox.bronze.v_customers
```

#### 6. Temporary View(valid only during spark session)
When a notebook is attached to cluster, spark session is active and temp view can be retrived 
If we detach and re-attach the cluster, temp view will be distroyed

```python
%sql
CREATE OR REPLACE TEMPORARY VIEW tv_customers AS
SELECT 
    *,
    _metadata.file_path
FROM json.`dbfs:/Volumes/gizmobox/landing/operational_data/customers/`;
```

```python
%sql
SELECT * FROM tv_customers
```

#### 7. Global temporary view
GTV wll be destroyed on the restarting of the cluster

```python
%sql
CREATE OR REPLACE GLOBAL TEMPORARY VIEW gtv_customers AS
SELECT 
    *,
    _metadata.file_path
FROM json.`dbfs:/Volumes/gizmobox/landing/operational_data/customers/`;
```

```python
%sql
SELECT * FROM global_temp.gtv_customers
```
