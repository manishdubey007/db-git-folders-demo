### Setting up the project environment for GizmoBox Data Lakehouse

#### 1. Access the container gizmobox

```python
%fs ls 'abfss://gizmobox@databrickscourseextmandl.dfs.core.windows.net/landing/operational_data/'
```

#### 2. Create external location

```python
%sql
CREATE EXTERNAL LOCATION IF NOT EXISTS dea_course_ext_dl_gizmobox
    URL 'abfss://gizmobox@databrickscourseextmandl.dfs.core.windows.net/'
    WITH (STORAGE CREDENTIAL dsa_course_ext_sc)
    COMMENT 'External Location for gizmobox Purposes;'
```

#### 3. Create the catalog - gizmobox

```python
%sql
SHOW CATALOGS;
```

```python
%sql
CREATE CATALOG IF NOT EXISTS gizmobox
      MANAGED LOCATION 'abfss://gizmobox@databrickscourseextmandl.dfs.core.windows.net/'
      COMMENT 'This is the catalog for the Gizmobox Data Lakehouse' ;
```

#### 4. Create schemas for
- Landing
- Bronze
- Silver
- Gold

```python
%sql
SELECT current_catalog();
```

```python
%sql
USE CATALOG gizmobox;

CREATE SCHEMA IF NOT EXISTS landing
       MANAGED LOCATION 'abfss://gizmobox@databrickscourseextmandl.dfs.core.windows.net/landing/';
CREATE SCHEMA IF NOT EXISTS bronze
       MANAGED LOCATION 'abfss://gizmobox@databrickscourseextmandl.dfs.core.windows.net/bronze/';
CREATE SCHEMA IF NOT EXISTS silver
       MANAGED LOCATION 'abfss://gizmobox@databrickscourseextmandl.dfs.core.windows.net/silver/';
CREATE SCHEMA IF NOT EXISTS gold
       MANAGED LOCATION 'abfss://gizmobox@databrickscourseextmandl.dfs.core.windows.net/gold/';
```

```python
%sql
SHOW SCHEMAS;
```

#### 5. Create Volume

```python
%sql
USE CATALOG gizmobox;
USE SCHEMA landing;

CREATE EXTERNAL VOLUME IF NOT EXISTS operational_data
    LOCATION 'abfss://gizmobox@databrickscourseextmandl.dfs.core.windows.net/landing/operational_data/';
```
