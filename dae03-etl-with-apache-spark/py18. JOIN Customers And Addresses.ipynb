####JOIN Customers And Addresses; Create customer_address table

```python
address_df = spark.read.table("gizmobox.silver.py_addresses")

customer_df = spark.read.table("gizmobox.silver.py_customers")
customer_address_df = address_df.join(customer_df, address_df.customer_id == customer_df.customer_id, "inner").select(
    customer_df.customer_id.alias("customer_id"),
    customer_df.customer_name.alias("customer_name"),
    customer_df.date_of_birth.alias("date_of_birth"),
    customer_df.email.alias("email"),
    customer_df.member_since.alias("member_since"),
    customer_df.telephone.alias("telephone"),
    address_df.shipping_address_line_1.alias("shipping_address_line_1"),
    address_df.shipping_city.alias("shipping_city"),
    address_df.shipping_state.alias("shipping_state"),
    address_df.shipping_postcode.alias("shipping_postcode"),
    address_df.billing_address_line_1.alias("billing_address_line_1"),
    address_df.billing_city.alias("billing_city"),
    address_df.billing_state.alias("billing_state"),
    address_df.billing_postcode.alias("billing_postcode")
    )
display(customer_address_df)
```

```python
customer_address_df.writeTo("gizmobox.gold.py_customer_address").createOrReplace()
```
