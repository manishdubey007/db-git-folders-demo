#### 4. Select file metadat

```python

df_with_metadata = spark.read.format("json").load("dbfs:/Volumes/gizmobox/landing/operational_data/customers/*").select("_metadata.file_path", "*")
display(df)

### https://spark.apache.org/docs/latest/api/python/reference/pyspark.sql/api/pyspark.sql.DataFrame.select.html
```

#### 5. Create Table in bronze schema

```python
df_with_metadata.write.mode("overwrite").format("delta").saveAsTable("gizmobox.bronze.py_customers")

### https://spark.apache.org/docs/latest/api/python/reference/pyspark.sql/api/pyspark.sql.DataFrameWriter.saveAsTable.html
```

```python
%sql
SELECT * FROM gizmobox.bronze.py_customers;
```

```python
df_with_metadata.writeTo("gizmobox.bronze.py_customers").createOrReplace() 
```
