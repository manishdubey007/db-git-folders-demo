#### Transform Addresses Data
- Create 1 record for each customer with 2 sets of address columns, 1 for shipping address and 1 for billing address
- Write transformed data in to the silvaer layer

#### 1. Create 1 record for each customer with 2 sets of address columns, 1 for shipping address and 1 for billing address
- https://docs.databricks.com/aws/en/sql/language-manual/sql-ref-syntax-qry-select-pivot
- https://spark.apache.org/docs/latest/api/python/reference/pyspark.sql/api/pyspark.sql.GroupedData.pivot.html


```python
from pyspark.sql import functions as F

df = spark.read.table("gizmobox.bronze.py_addresses")
extracted_df = df.groupBy("customer_id").pivot("address_type", ["billing", "shipping"]).agg(
            F.max("address_line_1").alias("address_line_1"), 
            F.max("city").alias("city"), 
            F.max("state").alias("state"), 
            F.max("postcode").alias("postcode")
        )
display(extracted_df)
```

#### 2. Write transformed data in to the silvaer layer

```python
extracted_df.writeTo("gizmobox.silver.py_addresses").createOrReplace()
```
