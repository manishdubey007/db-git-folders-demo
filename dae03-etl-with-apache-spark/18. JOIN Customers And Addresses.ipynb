####JOIN Customers And Addresses; Create customer_address table

```python
%sql
CREATE TABLE IF NOT EXISTS gizmobox.gold.customer_address
SELECT 
    c.customer_id,
    c.customer_name,
    c.email,
    c.date_of_birth,
    c.member_since,
    c.telephone,
    a.shipping_address_line_1,
    a.shipping_address_city,
    a.shipping_address_state,
    a.shipping_address_postcode,
    a.billing_address_line_1,
    a.billing_address_city,
    a.billing_address_state,
    a.billing_address_postcode
FROM gizmobox.silver.customers  c
JOIN gizmobox.silver.addresses a
ON c.customer_id = a.customer_id;
```
