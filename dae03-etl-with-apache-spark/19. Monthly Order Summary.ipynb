#### Monthly Order Summary for each customer per month

- Total orders
- Total items bought
- Total amount spent

#### 1. Total orders

```python
%sql
SELECT * FROM gizmobox.silver.orders
```

```python
%sql
SELECT
    customer_id,
    date_format(transaction_timestamp, 'yyyy-MM') AS order_month,
    COUNT(DISTINCT order_id)
FROM gizmobox.silver.orders
GROUP BY
    customer_id,
    order_month
ORDER BY
    customer_id,
    order_month;
```

####2. Total items bought


```python
%sql
SELECT
    customer_id,
    date_format(transaction_timestamp, 'yyyy-MM') AS order_month,
    SUM(quantity)
FROM gizmobox.silver.orders
GROUP BY
    customer_id,
    order_month
ORDER BY
    customer_id,
    order_month;
```

####3. Total amount spent

```python
%sql
SELECT
    date_format(transaction_timestamp, 'yyyy-MM') AS order_month,
    customer_id,
    COUNT(DISTINCT order_id) AS total_orders,
    SUM(quantity) AS total_items,
    SUM(quantity * price) AS total_amount
FROM gizmobox.silver.orders
GROUP BY
    order_month,
    customer_id
ORDER BY
    order_month,
    customer_id;
```

```python
%sql
WITH orders AS (
    SELECT
        DISTINCT
        customer_id,
        date_format(transaction_timestamp, 'yyyy-MM') AS order_month,
        order_id,
        total_amount
    FROM gizmobox.silver.orders
)
SELECT
    customer_id,
    order_month,
    SUM(total_amount)
FROM orders
GROUP BY
    customer_id,
    order_month
ORDER BY
    customer_id,
    order_month;
```

```python
%sql
CREATE TABLE IF NOT EXISTS gizmobox.gold.order_summary_monthly
AS
SELECT
    date_format(transaction_timestamp, 'yyyy-MM') AS order_month,
    customer_id,
    COUNT(DISTINCT order_id) AS total_orders,
    SUM(quantity) AS total_items,
    SUM(quantity * price) AS total_amount
FROM gizmobox.silver.orders
GROUP BY
    order_month,
    customer_id;
```
