#### 2. Create external table 

```python
%sql
CREATE TABLE IF NOT EXISTS gizmobox.bronze.payments
    (payment_id INTEGER,
    order_id INTEGER,
    payment_timstamp TIMESTAMP,
    payement_status INTEGER,
    payment_method STRING
    )
USING CSV
OPTIONS (
    header = "false",
    delimeter = ","
)
LOCATION "abfss://gizmobox@databrickscourseextmandl.dfs.core.windows.net/landing/external_data/payments/"
```

```python
%sql 
SELECT * FROM gizmobox.bronze.payments
```

```python
%sql
DESCRIBE EXTENDED gizmobox.bronze.payments
```

#### 3. Demo the effect of Adding/Updating/Deleting files

```python
%sql 
SELECT * FROM gizmobox.bronze.payments;
REFRESH TABLE gizmobox.bronze.payments;
```

####4. Demo the effect of dropping the table (only metdata will be dropped and data will remain as is)

```python
%sql
DROP TABLE IF EXISTS gizmobox.bronze.payments;
```
