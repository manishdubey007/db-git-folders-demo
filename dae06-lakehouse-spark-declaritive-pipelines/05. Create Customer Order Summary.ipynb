###Create Customer Order Summary
- Join the tables
- Retrieve latest address of the customer
- Calculate the following values
    - total orders
    - total items ordered
    - total order amount

####1. Join the tables

```sql
CREATE OR REFRESH MATERIALIZED VIEW gold_customer_order_summary 
AS 
SELECT 
    c.customer_id,
    c.customer_name,
    c.date_of_birth,
    c.telephone,
    c.email,
    a.address_line_1,
    a.city,
    a.state,
    a.postcode,
    COUNT(DISTINCT o.order_id) AS total_orders,
    SUM(o.quantity) AS total_items_ordered,
    SUM(o.quantity * o.price) AS  total_order_amount
FROM LIVE.silver_customers c
INNER JOIN LIVE.silver_orders o
    ON c.customer_id = o.customer_id
INNER JOIN LIVE.silver_addresses a
    ON c.customer_id = a.customer_id
WHERE a.__END_AT IS NULL
GROUP BY ALL;

```
