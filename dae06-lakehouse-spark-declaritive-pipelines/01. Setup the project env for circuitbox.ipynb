###Set up the project environment for CircuitBox Data LakeHouse
- Create External Location
- Create Catalog
- Create Schema
    - landing
    - lakehouse
- Create Volume

![image_1787811808791.png](./image_1787811808791.png "image_1787811808791.png")

#### 1. Create External Location
- Ext Loc : dea_course_ext_dl_circuitbox
- ADLS path : 'abfss://circuitbox@databrickscourseextmandl.dfs.core.windows.net/'
- Storage Credentials : dsa_course_ext_sc


```python
%sql
CREATE EXTERNAL LOCATION IF NOT EXISTS dea_course_ext_dl_circuitbox
    URL 'abfss://circuitbox@databrickscourseextmandl.dfs.core.windows.net/'
    WITH (STORAGE CREDENTIAL dsa_course_ext_sc)
    COMMENT 'External location for circuitbox data lakehouse';
```

```python
%fs ls 'abfss://circuitbox@databrickscourseextmandl.dfs.core.windows.net'
```

#### 2. Creating Catalog
- Catalog name : circuitbox
- Managed location: ['abfss://circuitbox@databrickscourseextmandl.dfs.core.windows.net/'](url)

```python
%sql
CREATE CATALOG IF NOT EXISTS circuitbox
    MANAGED LOCATION 'abfss://circuitbox@databrickscourseextmandl.dfs.core.windows.net/'
    COMMENT 'Catalog for the circuitbox data lakehouse';
    
```

```python
%sql
SHOW CATALOGS
```

#### 3. Creating Schema
- landing 
    - Managed Location: ['abfss://circuitbox@databrickscourseextmandl.dfs.core.windows.net/landing/'](url)
- lakehouse :
    - Managed Location: ['abfss://circuitbox@databrickscourseextmandl.dfs.core.windows.net/lakehouse'](url)

```python
%sql
USE CATALOG circuitbox;
    
CREATE SCHEMA IF NOT EXISTS landing
    MANAGED LOCATION 'abfss://circuitbox@databrickscourseextmandl.dfs.core.windows.net/landing/'
    COMMENT 'Schema for the landing data';
    
CREATE SCHEMA IF NOT EXISTS lakehouse
    MANAGED LOCATION 'abfss://circuitbox@databrickscourseextmandl.dfs.core.windows.net/lakehouse/'
    COMMENT 'Schema for the lakehouse';
```

```python
%sql
SHOW SCHEMAS;
```

#### 4. Create volume
- Volume Name : operational_data
- Managed Location : ['abfss://circuitbox@databrickscourseextmandl.dfs.core.windows.net/landing/operational_data'](url)

```python
%sql
USE CATALOG circuitbox;
USE SCHEMA landing;

CREATE EXTERNAL VOLUME IF NOT EXISTS operational_data
    LOCATION 'abfss://circuitbox@databrickscourseextmandl.dfs.core.windows.net/landing/operational_data/'
```
